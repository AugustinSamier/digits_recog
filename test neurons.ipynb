```python
import numpy as np
import copy
```

```python
x_train,y_train=np.zeros((10,10)),np.zeros((10,10))
for u in range(10):
    x_train[u][u]=1
    y_train[u][u]=1
print(x_train)
```

```python
weights,bias=[],[]
weights1=np.random.randn(10,10)
weights.append(weights1)
bias1=np.random.randn(10)
bias.append(bias1)
```

```python
def forward(x_train,y_train,weights,bias):
    outputs=[]
    outputs1=np.zeros((len(x_train),10))
    for data in range(len(x_train)):
        for neurone in range(weights[0].shape[0]):
            #outputs1[data][neurone]=((np.sum((weights[0][neurone]*x_train[data]))+bias[0][neurone]) if (np.sum((weights[0][neurone]*x_train[data]))+bias[0][neurone])>0 else 0)
            outputs1[data][neurone]=(np.sum((weights[0][neurone]*x_train[data]))+bias[0][neurone])
        outputs.append(outputs1)
    return outputs
```

```python
outputs=forward(x_train,y_train,weights,bias)
```

```python
def backprop(x_train,y_train,outputs,wei,bia,alpha=0.001):
    wei=copy.deepcopy(wei)
    bia=copy.deepcopy(bia)
    gradW1=np.zeros(wei[0].shape)
    gradsW1=np.zeros(wei[0].shape)
    gradsB1=np.zeros(bia[0].shape)
    gradMSE=np.zeros(wei[0].shape[0])
    for data in range(len(x_train)):
        for neurone in range(wei[0].shape[0]):
            #gradMSE[neurone]=(((outputs[0][data][neurone]-y_train[data][neurone])*2)/len(wei[0]) if (outputs[0][data][neurone])>0 else 0)
            gradMSE[neurone]=((outputs[0][data][neurone]-y_train[data][neurone])*2)/len(wei[0])
            for w in range(wei[0].shape[1]):
                gradW1[neurone][w]=gradMSE[neurone]*x_train[data][w]
        gradsB1+=gradMSE/len(x_train)
        gradsW1+=gradW1/len(x_train)

    wei[0]=wei[0]-alpha*gradsW1
    bia[0]=bia[0]-alpha*gradsB1
    return wei,bia
```

```python
def training(x_train,y_train,weights,bias,nb_epochs,alpha=0.001):
    we=copy.deepcopy(weights)
    bi=copy.deepcopy(bias)
    for i in range(nb_epochs):
        outputs=[]
        outputs=forward(x_train,y_train,we,bi)
        MSE=[]
        for data in range(len(x_train)):
            MSE.append(np.sum((outputs[0][data]-y_train[data])**2))
        print(f"MSE totale epoch {i+1}/{nb_epochs} : {np.sum(MSE)}")
        we,bi=backprop(x_train,y_train,outputs,we,bi,alpha)
    outputs=[]
    outputs=forward(x_train,y_train,we,bi)
    MSE=[]
    for data in range(len(x_train)):
        MSE.append(np.sum((outputs[0][data]-y_train[data])**2))
    print(f"MSE finale epoch : {np.sum(MSE)}")
    return we,bi
```

```python
wi,be=training(x_train,y_train,weights,bias,10000,alpha=0.01)
```

```python
outputstest=forward(x_train,y_train,wi,be)
true=0
for nb in range(10):
    print(f"Reel : {nb}, pred : {np.argmax(outputstest[0][nb])}\n")
    if nb==(np.argmax(outputstest[0][nb])):
        true+=1
print(true)
```
