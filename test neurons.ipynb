```python
import numpy as np
import copy
```

```python
x_train,y_train=np.zeros((10,10)),np.zeros((10,10))
for u in range(10):
    x_train[u][u]=1
    y_train[u][u]=1
print(x_train)
```

```python
weights,bias=[],[]
weights1=np.random.randn(10,10)
weights.append(weights1)
bias1=np.random.randn(10)
bias.append(bias1)
```

```python
def forward(x_train,y_train,weights,bias):
    outputs=[]
    outputs1=np.zeros((len(x_train),10))
    for data in range(len(x_train)):
        for neurone in range(weights[0].shape[0]):
            outputs1[data][neurone]=np.sum((weights[0][neurone]*x_train[data]))+bias[0][neurone]
        outputs.append(outputs1)
    return outputs
```

```python
outputs=forward(x_train,y_train,weights,bias)
```

```python
print(outputs[0][0])
```

```python
def backprop(x_train,y_train,outputs,wei,bia,alpha=0.001):
    wei=copy.deepcopy(wei)
    bia=copy.deepcopy(bia)
    MSE=np.zeros(len(x_train))
    gradW1=np.zeros(wei[0].shape)
    gradsW1=np.zeros(wei[0].shape)
    gradsB1=np.zeros(bia[0].shape)
    gradMSE=np.zeros(wei[0].shape[0])
    for data in range(len(x_train)):
        MSE[data]=(np.sum((outputs[0][data]-y_train[data])**2))/len(x_train[data])
        for neurone in range(wei[0].shape[0]):
            gradMSE[neurone]=((outputs[0][data][neurone]-y_train[data][neurone])*2)/len(wei[0])
            for w in range(wei[0].shape[1]):
                gradW1[neurone][w]=gradMSE[neurone]*wei[0][neurone][w]

        gradsB1+=gradMSE/len(x_train)
        gradsW1+=gradW1/len(x_train)

    wei[0]=wei[0]-alpha*gradW1
    bia[0]=bia[0]-alpha*gradsB1
    return wei,bia
```

```python
w,b=backprop(x_train,y_train,outputs,weights,bias)
print(f"Ancien w1 : \n{weights[0][0]}, \nnews:\n {w[0][0]}\n")
print(f"Ancien b1 : \n{bias[0][0]}, \nnews:\n {b[0][0]}")
```
